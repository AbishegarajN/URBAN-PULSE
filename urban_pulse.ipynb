```python
import matplotlib.pyplot as plt

status_counts = (
    df["status"]
    .fillna("Unspecified")
    .value_counts()
)

fig, ax = plt.subplots(figsize=(10, 7))

wedges, texts, autotexts = ax.pie(
    status_counts,
    labels=None,
    autopct=lambda p: f"{p:.1f}%" if p >= 2 else "",
    startangle=90,
    pctdistance=0.78,
    wedgeprops={"width": 0.4, "edgecolor": "white"},
    textprops={"fontsize": 11}
)

ax.text(
    0, 0,
    f"{status_counts.sum():,}\nComplaints",
    ha="center",
    va="center",
    fontsize=16,
    fontweight="bold"
)

ax.legend(
    wedges,
    [f"{status}: {count:,}" for status, count in status_counts.items()],
    title="Complaint Status",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)

ax.set_title(
    "NYC 311 Complaint Status Distribution",
    fontsize=16,
    fontweight="bold",
    pad=20
)

plt.tight_layout()
plt.show()
```

```python
agency_counts = df["agency"].value_counts().head(10)

plt.figure(figsize=(12, 7))

bars = plt.barh(
    agency_counts.index[::-1],
    agency_counts.values[::-1]
)

plt.bar_label(bars, padding=5, fmt="%.0f")

plt.title(
    "Top 10 NYC 311 Agencies by Complaint Volume",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Number of Complaints")
plt.ylabel("Agency")

plt.xlim(0, agency_counts.max() * 1.15)

plt.tight_layout()
plt.show()
```

```python
agency_performance = df.groupby("agency").agg(
    Total_Complaints=("unique_key", "count"),
    Closed_Complaints=("is_closed", "sum")
)

agency_performance["Closure_Rate"] = (
    agency_performance["Closed_Complaints"] /
    agency_performance["Total_Complaints"] * 100
)

agency_performance = agency_performance.sort_values(
    "Closure_Rate",
    ascending=True
).tail(10)

plt.figure(figsize=(12, 7))

bars = plt.barh(
    agency_performance.index,
    agency_performance["Closure_Rate"]
)

plt.bar_label(bars, fmt="%.1f%%", padding=5)

plt.title(
    "Top 10 Agencies by Complaint Closure Rate",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Closure Rate (%)")
plt.ylabel("Agency")

plt.xlim(0, 115)
plt.tight_layout()
plt.show()
```

```python
agency_open = df.groupby("agency").agg(
    Total_Complaints=("unique_key", "count"),
    Closed_Complaints=("is_closed", "sum")
)

agency_open["Open_Complaints"] = (
    agency_open["Total_Complaints"] -
    agency_open["Closed_Complaints"]
)

agency_open = agency_open.sort_values(
    "Open_Complaints",
    ascending=False
).head(10)

plt.figure(figsize=(12, 7))

bars = plt.barh(
    agency_open.index[::-1],
    agency_open["Open_Complaints"].values[::-1]
)

plt.bar_label(bars, padding=5, fmt="%.0f")

plt.title(
    "Top 10 Agencies by Outstanding Complaints",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Number of Open Complaints")
plt.ylabel("Agency")

plt.xlim(0, agency_open["Open_Complaints"].max() * 1.15)

plt.tight_layout()
plt.show()
```

```python
df["created_date"] = pd.to_datetime(
    df["created_date"],
    errors="coerce"
)

print("Data type:", df["created_date"].dtype)
print("Missing dates:", df["created_date"].isna().sum())
```

```python
daily_complaints = (
    df.groupby(df["created_date"].dt.date)
    .size()
)

plt.figure(figsize=(13, 6))

plt.plot(
    daily_complaints.index,
    daily_complaints.values,
    marker="o",
    linewidth=2
)

plt.title(
    "Daily NYC 311 Complaint Trend",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Date")
plt.ylabel("Number of Complaints")

plt.xticks(rotation=45)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
daily_status = df.groupby(
    df["created_date"].dt.date
).agg(
    Total_Complaints=("unique_key", "count"),
    Closed=("is_closed", "sum")
)

daily_status["Open"] = (
    daily_status["Total_Complaints"] -
    daily_status["Closed"]
)

plt.figure(figsize=(13, 6))

plt.plot(
    daily_status.index,
    daily_status["Closed"],
    marker="o",
    label="Closed Complaints"
)

plt.plot(
    daily_status.index,
    daily_status["Open"],
    marker="o",
    label="Open Complaints"
)

plt.title(
    "Daily Open vs Closed NYC 311 Complaints",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Date")
plt.ylabel("Number of Complaints")
plt.xticks(rotation=45)
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
daily_status["Closure_Rate"] = (
    daily_status["Closed"] /
    daily_status["Total_Complaints"] * 100
)

plt.figure(figsize=(13, 6))

plt.plot(
    daily_status.index,
    daily_status["Closure_Rate"],
    marker="o",
    linewidth=2
)

plt.axhline(
    y=67.34,
    linestyle="--",
    label="Overall Closure Rate (67.34%)"
)

plt.title(
    "Daily NYC 311 Complaint Closure Rate",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Date")
plt.ylabel("Closure Rate (%)")
plt.xticks(rotation=45)
plt.ylim(0, 100)
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
